## Frequent Itemset Mining using Apriori Algorithm

```python
# Install the mlxtend library for Apriori algorithm
!pip install mlxtend
```

```python
import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori

# Sample dataset of transactions
dataset = [['Milk', 'Onion', 'Eggs', 'Yogurt'],
           ['Dill', 'Onion', 'Eggs', 'Yogurt'],
           ['Milk', 'Apple', 'Eggs'],
           ['Milk', 'Unicorn', 'Corn','Yogurt'],
           ['Corn', 'Onion', 'Ice cream', 'Eggs']]

# Initialize TransactionEncoder
te = TransactionEncoder()

# Fit and transform the dataset into a one-hot encoded DataFrame
te_ary = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_ary, columns=te.columns_)

display(df.head())
```

Now, let's apply the Apriori algorithm to find frequent itemsets. We'll set a `min_support` threshold to filter itemsets that appear frequently enough.

```python
# Apply the Apriori algorithm to find frequent itemsets
frequent_itemsets = apriori(df, min_support=0.6, use_colnames=True)

# Display the frequent itemsets
print("Frequent Itemsets:")
display(frequent_itemsets)
```

Now that we have the frequent itemsets, we can generate association rules from them. We'll specify a `min_confidence` to filter for rules with a certain level of certainty.

```python
from mlxtend.frequent_patterns import association_rules

# Generate association rules with a minimum confidence of 0.7
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

# Display the association rules
print("Association Rules:")
display(rules)
```

### Frequent Pattern Growth (FP-Growth) Algorithm

While Apriori is a classic algorithm, FP-Growth is often more efficient for larger datasets as it doesn't generate candidate itemsets explicitly. It uses a tree-like data structure called an FP-tree to compress the dataset and then mines frequent itemsets directly from this tree.

```python
from mlxtend.frequent_patterns import fpgrowth

# Apply the FP-Growth algorithm to find frequent itemsets
# We'll use the same min_support as with Apriori for comparison
frequent_itemsets_fpgrowth = fpgrowth(df, min_support=0.6, use_colnames=True)

# Display the frequent itemsets found by FP-Growth
print("Frequent Itemsets (FP-Growth):")
display(frequent_itemsets_fpgrowth)
```
