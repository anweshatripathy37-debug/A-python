### Sample Data for Demonstration

Let's start by creating a sample DataFrame to demonstrate correlation and covariance analysis.

````python
import pandas as pd
import numpy as np

# Create a sample DataFrame
data = {
    'Feature_A': np.random.rand(100) * 10,
    'Feature_B': np.random.rand(100) * 5 + 2,
    'Feature_C': np.random.rand(100) * 8 - 3,
    'Feature_D': np.random.rand(100) * 15 + np.random.rand(100) * 5
}
df = pd.DataFrame(data)

# Introduce some correlation: Feature_E is positively correlated with Feature_A
df['Feature_E'] = df['Feature_A'] * 0.7 + np.random.randn(100) * 2

# Introduce some inverse correlation: Feature_F is negatively correlated with Feature_B
df['Feature_F'] = df['Feature_B'] * -0.5 + np.random.randn(100) * 1

print("Sample DataFrame head:")
display(df.head())
````

### Correlation Analysis

Correlation measures the strength and direction of a linear relationship between two variables. The correlation coefficient ranges from -1 to +1.

*   **+1:** Perfect positive linear relationship
*   **-1:** Perfect negative linear relationship
*   **0:** No linear relationship

````python
print("Correlation Matrix:")
correlation_matrix = df.corr()
display(correlation_matrix)
````

### Covariance Analysis

Covariance measures how two variables change together. A positive covariance indicates that variables tend to move in the same direction, while a negative covariance suggests they move in opposite directions. The magnitude of covariance is not standardized, making it harder to interpret the strength of the relationship compared to correlation.

````python
print("Covariance Matrix:")
covariance_matrix = df.cov()
display(covariance_matrix)
````

### Min-Max Normalization using the Formula (Manual Calculation)

Let's apply the Min-Max Normalization formula directly to the `df` DataFrame to see the process step-by-step.

````python
import pandas as pd

# Calculate x_min and x_max for each feature
x_min = df.min()
x_max = df.max()

# Apply the Min-Max Normalization formula
df_normalized_manual = (df - x_min) / (x_max - x_min)

print("DataFrame after Min-Max Normalization using the formula (first 5 rows):")
display(df_normalized_manual.head())
````

## Z-score Normalization Technique (Standardization)

Z-score normalization, also known as standardization, transforms the data such that it has a mean of 0 and a standard deviation of 1. This technique is particularly useful when the data follows a Gaussian distribution.

The formula for Z-score normalization for a given value `x` is:

```
x_standardized = (x - mean(x)) / std(x)
```

Where:
*   `x_standardized` is the standardized value.
*   `x` is the original value.
*   `mean(x)` is the mean of the feature.
*   `std(x)` is the standard deviation of the feature.

Standardization is often preferred over Min-Max scaling when the data contains outliers, as it will not scale them to a small predetermined range, thus preserving useful information about the outliers. It's also a common requirement for many machine learning algorithms like Linear Regression, Logistic Regression, and K-Means clustering, which assume that the data is normally distributed or that features have similar variances.

````python
from sklearn.preprocessing import StandardScaler
import pandas as pd

# Initialize the StandardScaler
scaler_zscore = StandardScaler()

# Apply Z-score Normalization to the DataFrame
# Similar to MinMaxScaler, fit_transform returns a numpy array, which we convert back to a DataFrame.
df_zscore_normalized = pd.DataFrame(scaler_zscore.fit_transform(df), columns=df.columns)

print("DataFrame after Z-score Normalization (first 5 rows):")
display(df_zscore_normalized.head())
````

## Decimal Scaling Normalization Technique

Decimal Scaling is a simple normalization technique that scales data by moving the decimal point of values. This method preserves the original data distribution and is particularly useful when you want to keep all values within a specific range, typically between -1 and 1.

The formula for Decimal Scaling for a given value `x` is:

```
x_scaled = x / 10^j
```

Where:
*   `x_scaled` is the scaled value.
*   `x` is the original value.
*   `j` is the smallest integer such that the maximum absolute value of `x` is less than 1 (i.e., `max(|x_i|) < 10^j`).

This technique is straightforward and ensures that all data points fall within a range of -1 to 1, or 0 to 1 if all values are positive. It's less sensitive to outliers than Min-Max scaling, but it doesn't standardize the data to a mean of 0 and a standard deviation of 1 like Z-score normalization.

````python
import numpy as np
import pandas as pd

# Calculate the maximum absolute value in the DataFrame
max_abs_value = df.abs().max().max()

# Determine 'j', the smallest integer such that max_abs_value < 10^j
j = np.ceil(np.log10(max_abs_value)).astype(int)

# Apply Decimal Scaling
df_decimal_scaled = df / (10**j)

print(f"The maximum absolute value in the original DataFrame is: {max_abs_value:.2f}")
print(f"The scaling factor (10^j) used is: {10**j}")
print("DataFrame after Decimal Scaling (first 5 rows):")
display(df_decimal_scaled.head())
````

As shown above, all values in the `df_decimal_scaled` DataFrame are now scaled by a power of 10, ensuring that their absolute values are less than 1.
