```python
import numpy as np
from scipy import stats

# Example dataset
data = np.array([15, 20, 25, 30, 35, 40, 40, 45, 50, 55])
print("Original Data:", data)
```

```python


# Calculate Mean
mean_value = np.mean(data)
print(f"Mean: {mean_value:.2f}")

```

```python
# Calculate Median
median_value = np.median(data)
print(f"Median: {median_value:.2f}")

```

```python


# Calculate Mode
# `stats.mode` returns a ModeResult object, where .mode is the actual mode value
# and .count is the number of times it appears.
mode_result = stats.mode(data, keepdims=False)
print(f"Mode: {mode_result.mode}")

```

```python

# Calculate Standard Deviation
# ddof=0 for population standard deviation (default in numpy)
# ddof=1 for sample standard deviation
std_dev = np.std(data, ddof=0)
print(f"Standard Deviation (Population): {std_dev:.2f}")

```

```python


# Calculate Variance
# ddof=0 for population variance (default in numpy)
# ddof=1 for sample variance
variance = np.var(data, ddof=0)
print(f"Variance (Population): {variance:.2f}")
```

```python
import numpy as np
from scipy import stats

def calculate_statistics(data):
  mean=np.mean(data)
  median=np.median(data)
  mode=stats.mode(data, keepdims=False)
  std_dev=np.std(data)
  variance=np.var(data)
  print(f"Mean: {mean:.2f}")
  print(f"Median: {median:.2f}")
  print(f"Mode: {mode.mode}")
  print(f"Standard Deviation: {std_dev:.2f}")
  print(f"Variance: {variance:.2f}")

# Example dataset
data = np.array([15,20,25,30,35,40,40,45,50,55])
print("Original Data:", data)

# Call the function
calculate_statistics(data)
```
