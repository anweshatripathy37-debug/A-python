## K-Means Clustering

K-means is a popular unsupervised machine learning algorithm used for partitioning a dataset into a set of 'k' distinct, non-overlapping subgroups (clusters). The algorithm works by iteratively trying to minimize the within-cluster sum of squares. It requires the number of clusters 'k' to be specified beforehand.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

# Generate some synthetic data for demonstration
X, y = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# Apply K-Means clustering
kmeans = KMeans(n_clusters=4, init='k-means++', max_iter=300, random_state=0, n_init=10)
y_kmeans = kmeans.fit_predict(X)

# Plot the results
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(X[:, 0], X[:, 1], c=y_kmeans, s=50, cmap='viridis', alpha=0.7)

# Plot the cluster centers
centers = kmeans.cluster_centers_
ax.scatter(centers[:, 0], centers[:, 1], c='red', s=200, alpha=0.9, marker='X', label='Centroids')

ax.set_title('K-Means Clustering')
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.legend()
plt.show()
```

## K-Medoids Clustering (PAM - Partitioning Around Medoids)

K-medoids is another partitioning clustering algorithm similar to K-means, but it uses actual data points (medoids) as cluster centers instead of centroids (which can be hypothetical points). This makes K-medoids more robust to noise and outliers compared to K-means. It minimizes the sum of dissimilarities between points and their assigned medoids. We'll use `sklearn_extra` for this.

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn_extra.cluster import KMedoids

# Generate some synthetic data (using the same data for consistency)
X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

# Apply K-Medoids clustering
kmedoids = KMedoids(n_clusters=4, random_state=0)
y_kmedoids = kmedoids.fit_predict(X)

# Plot the results
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(X[:, 0], X[:, 1], c=y_kmedoids, s=50, cmap='viridis', alpha=0.7)

# Plot the medoids
medoids = kmedoids.cluster_centers_
ax.scatter(medoids[:, 0], medoids[:, 1], c='red', s=200, alpha=0.9, marker='X', label='Medoids')

ax.set_title('K-Medoids Clustering')
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.legend()
plt.show()
```
